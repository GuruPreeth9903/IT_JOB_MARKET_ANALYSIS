### Cleaning Source Column

```python
df["Source"] = df["Source"].str.replace("via ","")
```

### Creating Min_Exp and Max_Exp from Exp column

```python
df["Experience"].str.extract("(\d+)")
```

```python
df["Min_Exp"] = df["Experience"].str.extract("(\d+)")
```

```python
df["Max_Exp"] = df["Experience"].str.extract("-(\d+)")
```

### Converting Exp columns to numerical

```python
## Converting Exp columns to numerical
df["Min_Exp"] = pd.to_numeric(df["Min_Exp"])
df["Max_Exp"] = pd.to_numeric(df["Max_Exp"])
```

### Cleaning Text Column

```python
df["Company"] = df["Company"].str.strip()
df["Role"] = df["Role"].str.strip()
df["Location"] = df["Location"].str.strip()
```

### Handling missing values in Salary Column

```python
df["Salary"].unique()
```

```python
### Removing symbols
df["Salary"] = df["Salary"].astype(str)
df["Salary"] = df["Salary"].str.replace("₹","")
df["Salary"] = df["Salary"].str.replace("L/yr","")
```

```python
# Handling salary ranges and replacing 
sal_range = df["Salary"].str.extract(r'(\d+\.?\d*)\s*-\s*(\d+\.?\d*)')
sal_range = sal_range.astype(float)
df["Salary"] = sal_range.mean(axis=1)
```

### Handling missing values in Salary Column

```python
df["Salary"] = pd.to_numeric(df["Salary"])
```

```python
# Fill salary based on skills
df["Salary"] = df["Salary"].fillna(
    df.groupby(["Role","Min_Exp"])["Salary"].transform("mean")
)
```

```python
df["Salary"].isna().sum()
```

```python
# Fill salary based on skills
df["Salary"] = df["Salary"].fillna(
    df.groupby("Skills")["Salary"].transform("mean")
)
```

```python
df["Salary"].isna().sum()
```

```python
# Fill remaining values using experience
df["Salary"] = df["Salary"].fillna(
    df.groupby("Min_Exp")["Salary"].transform("mean"))
```

```python
df["Salary"].isna().sum()
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
```

```python
## Fixsing missing values in exp columns
df["Min_Exp"] = df["Min_Exp"].fillna(df["Min_Exp"].median())
df["Max_Exp"] = df["Max_Exp"].fillna(df["Max_Exp"].median())
```
